```python
X= df.drop(columns = ['K(MPa)' , 'epsilon_0' , 'n'], axis=1)
y_1 = df['K(MPa)'] 
```

```python
X1_train, X1_test, y1_train, y1_test = train_test_split(X, y_1, test_size=0.15, random_state=42)
```

```python
model1 =  LinearRegression()
```

```python
model1.fit(X1_train, y1_train)
```

```python
X_train_df = pd.DataFrame(X1_train)
X_test_df = pd.DataFrame(X1_test, columns=X_train_df.columns)
    
y_pred1 = model1.predict(X_test_df)
y_pred2 = model2.predict(X_test_df)
y_pred3 = model3.predict(X_test_df)
y_pred4 = model4.predict(X_test_df)

y_pred1 = y_pred1.flatten()
y_pred2 = y_pred2.flatten()
y_pred3 = y_pred3.flatten()
y_pred4 = y_pred4.flatten()


r2_1 = r2_score(y1_test, y_pred1)
r2_2 = r2_score(y1_test, y_pred2)
r2_3 = r2_score(y1_test, y_pred3)
r2_4 = r2_score(y1_test, y_pred4)


print(r2_1 , r2_2 , r2_3, r2_4)
```

```python
import statsmodels.api as sm
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split


# Séparer variables explicatives et variable cible
X = df.drop(columns=['K(MPa)', 'epsilon_0', 'n'])
y_1 = df['K(MPa)']

# Diviser en jeu d'entraînement et de test
X1_train, X1_test, y1_train, y1_test = train_test_split(X, y_1, test_size=0.15, random_state=42)

# 1. Ajouter constante pour intercept
X_train_const = sm.add_constant(X1_train)
X_test_const = sm.add_constant(X1_test)

# 2. Ajuster le modèle OLS
model = sm.OLS(y1_train, X_train_const).fit()

# 3. Prédictions avec intervalles de confiance pour X_test
predictions = model.get_prediction(X_test_const)

# 4. Résumé avec intervalles à 95%
summary = predictions.summary_frame(alpha=0.05)

# 5. Afficher résultats pour les 3 premiers échantillons test
for i in range(min(3, len(summary))):
    print(f"Sample {i+1}:")
    print(f" Prédiction moyenne: {summary['mean'].iloc[i]:.3f}")
    print(f" Intervalle de confiance (moyenne): [{summary['mean_ci_lower'].iloc[i]:.3f}, {summary['mean_ci_upper'].iloc[i]:.3f}]")
    print(f" Intervalle de prédiction: [{summary['obs_ci_lower'].iloc[i]:.3f}, {summary['obs_ci_upper'].iloc[i]:.3f}]")
    print()
```
